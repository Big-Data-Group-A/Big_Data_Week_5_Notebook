```python
import numpy as np
import pandas as pd
import os

if os.path.exists("week5_kigali_weather.csv"):
    df = pd.read_csv("week5_kigali_weather.csv")
else:
    df = pd.read_excel("week5_kigali_weather.xlsx")
    # Excel stores dates as serial numbers -> convert to YYYY-MM-DD text
    if np.issubdtype(df["date"].dtype, np.number):
        df["date"] = pd.to_datetime(df["date"], unit="D", origin="1899-12-30").dt.strftime("%Y-%m-%d")

dates = df["date"].astype(str).to_numpy()
temps = df["temp_c"].to_numpy(dtype=float)
rain  = df["rainfall_mm"].to_numpy(dtype=float)
hum   = df["humidity_pct"].to_numpy(dtype=float)
print(len(dates), "days loaded:", dates[0], "to", dates[-1])
```

```python
temps_list = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1]   # given for Part 1
raw = ["24.4", "25.9", "28.6", "19.3"]               # given for Ex 1.3
```

```python
fahrenheit = [t * 9/5 + 32 for t in temps_list]
print([round(f, 1) for f in fahrenheit])
```

```python
hot = [t for t in temps_list if t >= 26]
cool_rounded = [round(t) for t in temps_list if t < 24]
print("hot:", hot)
print("cool_rounded:", cool_rounded)
```

```python
raw_floats = [float(x) for x in raw]
print(raw_floats)
print("Average:", sum(raw_floats) / len(raw_floats))
```

```python
results = [("Aline", 85), ("Eric", 78), ("Diane", 91), ("Samuel", 92), ("Grace", 76)]
ranked = sorted(results, key=lambda r: r[1], reverse=True)
print("Top 3:", ranked[:3])
```

```python
rain_log = [(3, 0.0), (7, 12.5), (11, 3.2), (14, 26.2), (18, 8.9)]
wettest = max(rain_log, key=lambda p: p[1])
print("Wettest day:", wettest[0], "with", wettest[1], "mm")
```

```python
import time
n = 1_000_000
data = list(range(n))
arr = np.arange(n)

t0 = time.perf_counter()
doubled_list = [x * 2 for x in data]      # plain Python
t_list = time.perf_counter() - t0

t0 = time.perf_counter()
doubled_arr = arr * 2                      # NumPy
t_np = time.perf_counter() - t0

print(f"Python: {t_list:.4f}s | NumPy: {t_np:.5f}s | NumPy ~{t_list / t_np:.0f}x faster")
# Comment: NumPy was roughly 10-100x faster on my machine (your number will differ).
# It runs the loop in compiled C code instead of the Python interpreter.
```

```python
print(f"Mean: {temps.mean():.2f} °C")
print(f"Max:  {temps.max():.1f} °C")
print(f"Min:  {temps.min():.1f} °C")
print(f"Std:  {temps.std():.2f} °C")
```

```python
pos = np.argmax(temps)
print(f"Hottest day: {dates[pos]} at {temps[pos]} °C")
```

```python
hot_mask = temps >= 26
print("(a) Number of hot days:", hot_mask.sum())
print(f"(b) Average temperature on hot days: {temps[hot_mask].mean():.2f} °C")
```

```python
rainy_mask = rain > 0
print("(a) Rainy days:", rainy_mask.sum())
print(f"(b) Total rainfall: {rain.sum():.1f} mm")
print(f"(c) Heaviest single-day rainfall: {rain.max():.1f} mm")
```

```python
dry_mask = rain == 0
print(f"(a) Avg temp on rainy days: {temps[rainy_mask].mean():.2f} °C")
print(f"    Avg temp on dry days:   {temps[dry_mask].mean():.2f} °C")
print("(b) Hot AND rainy days:", ((temps >= 26) & (rain > 0)).sum())
print("(c) Days with humidity >= 70:", (hum >= 70).sum())
```

```python
anomaly = temps - temps.mean()
print("(a) Days more than 3 °C above the mean:")
print(dates[anomaly > 3])
print("(b) Days more than 3 °C below the mean:", (anomaly < -3).sum())
```

```python
first30 = temps[:30].mean()
last30  = temps[30:].mean()
print(f"First 30 days: {first30:.2f} °C")
print(f"Last 30 days:  {last30:.2f} °C")
print(f"Difference:    {last30 - first30:+.2f} °C")
# Comment: see the printed difference. Positive = Kigali got warmer, negative = cooler.
```

```python
top5 = sorted(zip(dates, rain), key=lambda p: p[1], reverse=True)[:5]
for d, r in top5:          # loop allowed here: printing the bonus
    print(d, "-", r, "mm")
```
