```python

import numpy as np

with open("week5_kigali_weather.csv") as f:
    lines = f.readlines()[1:]  # Skip the header

rows = [line.strip().split(",") for line in lines]

dates = np.array([row[1] for row in rows])
temps = np.array([float(row[2]) for row in rows])
rain = np.array([float(row[3]) for row in rows])
hum = np.array([float(row[4]) for row in rows])

print("Number of days:", temps.size)
print("First five temperatures:", temps[:5])
```

```python
print(f"Mean: {temps.mean():.2f} °C")
print(f"Maximum: {temps.max():.2f} °C")
print(f"Minimum: {temps.min():.2f} °C")
print(f"Standard deviation: {temps.std():.2f} °C")
```

```python
print(f"Mean: {temps.mean():.2f} °C")
print(f"Maximum: {temps.max():.2f} °C")
print(f"Minimum: {temps.min():.2f} °C")
print(f"Standard deviation: {temps.std():.2f} °C")
```

```python
position = np.argmax(temps)

print("Hottest day:", dates[position])
print("Highest temperature:", temps[position], "°C")
```

```python
hot_mask = temps >= 26

hot_temps = temps[hot_mask]

print("Number of hot days:", hot_mask.sum())
print(f"Average temperature on hot days: {hot_temps.mean():.2f} °C")
```

```python
rainy_mask = rain > 0

rainy_amounts = rain[rainy_mask]

print("Number of rainy days:", rainy_mask.sum())
print(f"Total rainfall: {rainy_amounts.sum():.2f} mm")
print(f"Heaviest rainfall: {rainy_amounts.max():.2f} mm")
```

```python
# Average temperature on rainy days
rainy_temps = temps[rain > 0]
print(f"Average temperature on rainy days: {rainy_temps.mean():.2f} °C")

# Average temperature on dry days
dry_temps = temps[rain == 0]
print(f"Average temperature on dry days: {dry_temps.mean():.2f} °C")

# Count days that were both hot and rainy
hot_and_rainy = (temps >= 26) & (rain > 0)
print("Number of hot and rainy days:", hot_and_rainy.sum())

# Count days with humidity of 70% or higher
humid_days = hum >= 70
print("Days with humidity >= 70%:", humid_days.sum())
```

```python
# Calculate how far each temperature is from the mean
anomaly = temps - temps.mean()

# Find days more than 3°C above the average
print("Days more than 3°C above average:")
print(dates[anomaly > 3])

# Count days more than 3°C below the average
print("Number of days more than 3°C below average:")
print((anomaly < -3).sum())
```

```python
first_half = temps[:30]
second_half = temps[30:]

first_avg = first_half.mean()
second_avg = second_half.mean()

print(f"First 30 days average: {first_avg:.2f} °C")
print(f"Last 30 days average: {second_avg:.2f} °C")

if second_avg > first_avg:
    print("The second half was warmer.")
elif second_avg < first_avg:
    print("The second half was cooler.")
else:
    print("Both halves had the same average temperature.")
```

```python
weather_log = list(zip(dates, rain))

wettest_days = sorted(
    weather_log,
    key=lambda day: day[1],
    reverse=True
)

print("Top 5 wettest days:")
for day in wettest_days[:5]:
    print(f"Date: {day[0]}, Rainfall: {day[1]:.2f} mm")
```
