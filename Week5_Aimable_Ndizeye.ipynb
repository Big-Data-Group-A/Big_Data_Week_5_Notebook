```python
with open("week5_kigali_weather.csv","r") as f:
    lines = f.readlines()

rows = [line.strip().split(",") for line in lines[1:]] 
dates =np.array ([r[1] for r in rows])
temps = np.array ([float(r[2]) for r in rows])
rain = np.array ([float(r[3]) for r in rows])
hum = np.array ([float(r[4]) for r in rows])
```

```python
## Exercises 3.3

pos = np.argmax(temps)
print (f"Hottest : {dates[pos]} with {temps[pos]} °C")
```

```python
## part 4 
## Exercises 4.1

hot = temps >= 26
##print (temps[hot])

print(f"a.The Number of hot days: {hot.sum()}");
print (f"b. The average on the hot days: {temps[hot].mean():.2f}")


```

```python
## Exercises 4.2 

Rainy = rain >0

##print(rain [cold])

print(f"(a).The Number of a rain days : {Rainy.sum()}")
print(f"(b).Total Rainfall: {rain.sum():.2f}")
print(f"(c).The heaviest single-day: {rain.max()}")
```

```python
## Exercises 4.3

dry = rain == 0

print(f"(a) Avg temp on rainy days: {temps[Rainy].mean():.2f} °C")
print(f"(b) Avg temp on dry days:   {temps[dry].mean():.2f} °C")
print(f"(c) Hot AND rainy days: {((temps >= 26) & (rain > 0)).sum()}")
print(f"(d) Days with humidity >= 70: {(hum >= 70).sum()}")
```

```python
## Part 5 Anomaly and Trends 
##  Exercise 5.1 

anomaly = temps - temps.mean()
print("(a) Days more than 3 °C above the mean:")
print(dates[anomaly > 3])
print(f"(b) Days more than 3 °C below the mean: {(anomaly < -3).sum()}")


```

```python
## Exercises 5.2

first30 = temps[:30].mean()
last30  = temps[30:].mean()
print(f"First 30 days average: {first30:.2f} °C")
print(f"Last 30 days average:  {last30:.2f} °C")
print(f"Difference: {last30 - first30:+.2f} °C")
# Comment: Kigali got COOLER. The last 30 days (October) averaged about 4.4 °C
# lower than the first 30 days (September).
```

```python
## Bonus  :  Top 5 wettest days

top5 = sorted(zip(dates, rain), key=lambda pair: pair[1], reverse=True)[:5]
for date, mm in top5:         
    print(f"{date}: {mm} mm")
```

```python
## Reflections 
# 1. What did NumPy hide from you, and why learn the loop version first?

# In Week 3, averaging a column meant opening the data, creating a running total and a counter, looping over every value, 
# adding each one, and dividing at the end. `temps.mean()` hides all of that: the loop, the accumulator, the counter and the
# division now happen inside fast compiled C code. The same applies to masks like `temps[temps >= 26]`, which hide the loop-plus-`if`
# filtering

# 2. One insight about Kigali's weather

# Kigali cooled sharply through the period: the first 30 days averaged 26.2 °C and the last 30 averaged 21.8 °C, 
# a drop of about 4.4 °C. Temperature and humidity moved in opposite directions (hot days were drier), and rain fell on only 16 of 60 days, 
# but always heavily (15–26 mm when it came), for 331 mm in total.
```
