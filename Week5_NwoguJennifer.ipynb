```python
# Exercise 3.1 - Speed test: Python loop vs NumPy
import numpy as np
import time

numbers = list(range(1_000_000))
arr = np.array(numbers)

# Pure Python loop
t0 = time.time()
doubled = [n * 2 for n in numbers]
python_time = time.time() - t0

print(f"Python: {python_time:.3f} s")

# NumPy vectorized operation
t0 = time.time()
doubled = arr * 2
numpy_time = time.time() - t0

print(f"NumPy: {numpy_time:.3f} s")

# Calculate how many times faster NumPy was
print(f"NumPy was approximately {python_time / numpy_time:.1f}x faster.")
```

```python
# Exercise 3.2 - Load the Kigali weather dataset

import numpy as np

# Open the CSV file
with open("week5_kigali_weather.csv") as f:
    lines = f.readlines()[1:]  # Skip the header row

# Split each row into columns
rows = [line.strip().split(",") for line in lines]

# Create NumPy arrays
temps = np.array([float(row[2]) for row in rows])
rain = np.array([float(row[3]) for row in rows])

# Check that all 60 days were loaded
print("Number of days:", temps.size)
```

```python
# Exercise 3.3 - Calculate basic weather statistics

# Calculate the mean, maximum, minimum, and standard deviation
print("Mean temperature:", round(np.mean(temps), 2), "°C")
print("Maximum temperature:", np.max(temps), "°C")
print("Minimum temperature:", np.min(temps), "°C")
print("Standard deviation:", round(np.std(temps), 2), "°C")
```

```python
# Exercise 3.4 - Find the hottest day

# Find the position of the highest temperature
hottest_index = np.argmax(temps)

# Get the hottest temperature
hottest_temp = temps[hottest_index]

# Get the date from the original rows
hottest_date = rows[hottest_index][1]

print("Hottest date:", hottest_date)
print("Hottest temperature:", hottest_temp, "°C")
```

```python
# Exercise 4.1 - Find hot days

# Create a Boolean mask for temperatures of 26°C or higher
hot_mask = temps >= 26

# Count the hot days
hot_days = np.sum(hot_mask)

# Calculate the average temperature on hot days
hot_average = np.mean(temps[hot_mask])

print("Number of hot days:", hot_days)
print("Average hot-day temperature:", round(hot_average, 2), "°C")
```

```python
# Exercise 4.2 - Find rainy days

# Create a Boolean mask for days with rainfall
rain_mask = rain > 0

# Count rainy days
rainy_days = np.sum(rain_mask)

# Calculate total and maximum rainfall
total_rain = np.sum(rain[rain_mask])
max_rain = np.max(rain[rain_mask])

print("Number of rainy days:", rainy_days)
print("Total rainfall:", round(total_rain, 1), "mm")
print("Maximum daily rainfall:", max_rain, "mm")
```

```python
# Exercise 4.3 - Compare rainy and dry days

# Calculate average temperature on rainy days
rainy_avg = np.mean(temps[rain_mask])

# Calculate average temperature on dry days
dry_avg = np.mean(temps[~rain_mask])

print("Average rainy-day temperature:", round(rainy_avg, 2), "°C")
print("Average dry-day temperature:", round(dry_avg, 2), "°C")
```

```python
# Exercise 4.4 - Find days that are both hot and rainy

# Create a mask for days that are both hot and rainy
hot_rain_mask = (temps >= 26) & (rain > 0)

# Count the hot and rainy days
hot_rain_days = np.sum(hot_rain_mask)

print("Number of hot and rainy days:", hot_rain_days)
```

```python
# Exercise 4.5 - Find days with high humidity

# Create the humidity array
hum = np.array([float(row[4]) for row in rows])

# Create a mask for humidity of 70% or higher
high_humidity_mask = hum >= 70

# Count the high-humidity days
high_humidity_days = np.sum(high_humidity_mask)

print("Number of high-humidity days:", high_humidity_days)
```

```python
# Exercise 5.1 - Find temperature anomalies

# Calculate the average temperature
mean_temp = np.mean(temps)

# Find days more than 3°C above the average
above_anomaly = temps > mean_temp + 3

# Find days more than 3°C below the average
below_anomaly = temps < mean_temp - 3

print("Days more than 3°C above average:")
print(np.array([rows[i][1] for i in np.where(above_anomaly)[0]]))

print("\nDays more than 3°C below average:")
print(np.array([rows[i][1] for i in np.where(below_anomaly)[0]]))
```

```python
# Exercise 5.2 - Compare the first 30 days with the last 30 days

# Calculate the average temperature for the first 30 days
first_30_avg = np.mean(temps[:30])

# Calculate the average temperature for the last 30 days
last_30_avg = np.mean(temps[30:])

print("First 30 days average:", round(first_30_avg, 2), "°C")
print("Last 30 days average:", round(last_30_avg, 2), "°C")

# Describe the temperature trend
if last_30_avg < first_30_avg:
    print("Trend: Kigali got cooler.")
else:
    print("Trend: Kigali got warmer.")
```

```python
# Bonus - Find the 5 wettest days

# Get the positions of the 5 highest rainfall values
wettest_indices = np.argsort(rain)[-5:][::-1]

# Print the 5 wettest days
print("Top 5 wettest days:")

for i in wettest_indices:
    print(rows[i][1], "-", rain[i], "mm")
```
