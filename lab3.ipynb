```python
# Importing necessary libraries and dataset
import numpy as np
import pandas as pd

df = pd.read_excel("week5_kigali_weather.xlsx")
df["date"] = pd.to_datetime(df["date"], unit="D", origin="1899-12-30")
dates = df["date"].to_numpy()
temps = df["temp_c"].to_numpy()
rain = df["rainfall_mm"].to_numpy()
hum = df["humidity_pct"].to_numpy()

results = list(zip(dates, temps))
rain_log = list(zip(dates, rain))

print(df.head())
```

```python
# Question 1.1 — Transforming temperatures from Celsius to Fahrenheit

sample_temps = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1]

fahrenheit = [t * 9/5 + 32 for t in sample_temps]

print(fahrenheit)
```

```python
# Question 1.2 — Filtering temperatures

hot = [t for t in sample_temps if t >= 26]
cool_rounded = [round(t) for t in sample_temps if t < 24]

print("Hot:", hot)
print("Cool rounded:", cool_rounded)
```

```python
# Question 1.3 — Convert types
raw = ["24.4", "25.9", "28.6", "19.3"]

raw_floats = [float(t) for t in raw]

print(raw_floats)
print("Average:", sum(raw_floats) / len(raw_floats))
```

```python
# Question 2.1 — Rank with sorted

ranked = sorted(results, key=lambda x: x[1], reverse=True)

print(ranked[:3])
```

```python
# Question 2.2 — max with a key

wettest = max(rain_log, key=lambda x: x[1])

print(wettest)
```

```python
# Question 3.1 — The speed test
import time
import numpy as np

arr = np.arange(1_000_000)

start = time.time()
loop_result = [x * 2 for x in arr]
loop_time = time.time() - start

start = time.time()
numpy_result = arr * 2
numpy_time = time.time() - start

print("Python loop:", loop_time, "seconds")
print("NumPy:", numpy_time, "seconds")
print("NumPy was approximately", loop_time / numpy_time, "times faster.")

# NumPy was approximately ___ times faster on my machine.
```

```python
# Question 3.2 — Describe the temperatures

print(f"Mean: {temps.mean():.2f} °C")
print(f"Maximum: {temps.max():.2f} °C")
print(f"Minimum: {temps.min():.2f} °C")
print(f"Standard deviation: {temps.std():.2f} °C")
```

```python
# Question 3.3 — The hottest day

position = np.argmax(temps)

print(f"Hottest day: {dates[position]}")
print(f"Temperature: {temps[position]:.2f} °C")
```

```python
# Question 4.1 — Hot days
hot_mask = temps >= 26

print("Number of hot days:", hot_mask.sum())
print(f"Average temperature on hot days: {temps[hot_mask].mean():.2f} °C")
```

```python
# Question 4.2 — Rain report
rain_mask = rain > 0

print("Number of rainy days:", rain_mask.sum())
print(f"Total rainfall: {rain.sum():.2f} mm")
print(f"Heaviest single-day rainfall: {rain.max():.2f} mm")
```

```python
# Question 4.3 — Cross-array analysis

rainy_mask = rain > 0
dry_mask = rain == 0

print(f"Average temperature on rainy days: {temps[rainy_mask].mean():.2f} °C")
print(f"Average temperature on dry days: {temps[dry_mask].mean():.2f} °C")

hot_and_rainy = (temps >= 26) & (rain > 0)

print("Hot AND rainy days:", hot_and_rainy.sum())
print("Days with humidity >= 70%:", (hum >= 70).sum())
```

```python
# Question 5.1 — Anomaly detection
anomaly = temps - temps.mean()

above_mask = anomaly > 3
below_mask = anomaly < -3

print("Dates more than 3 °C above the mean:")
print(dates[above_mask])

print("Days more than 3 °C below the mean:", below_mask.sum())
```

```python
# Question 5.2 — Is the season changing?
first_30_avg = temps[:30].mean()
last_30_avg = temps[30:].mean()

print(f"First 30 days average: {first_30_avg:.2f} °C")
print(f"Last 30 days average: {last_30_avg:.2f} °C")

# If the last 30-day average is higher, Kigali got warmer.
# If the last 30-day average is lower, Kigali got cooler.
```

```python
# Bonus — Top 5 wettest days
wettest_days = sorted(
    zip(dates, rain),
    key=lambda x: x[1],
    reverse=True
)

print("Top 5 wettest days:")

for date, rainfall in wettest_days[:5]:
    print(date, f"{rainfall:.2f} mm")
```
