Part 1 · List Comprehensions (6 marks)


```python
temps_list = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1]
raw = ["24.4", "25.9", "28.6", "19.3"]

# 1.1 Fahrenheit
temps_f = [t * 9/5 + 32 for t in temps_list]
print(temps_f)          # [75.92, 78.62, 83.48, 66.74, 78.8, 71.78]

```

```python

# 1.2 Filter
hot = [t for t in temps_list if t >= 26]
print(hot)              # [28.6, 26.0]
cool_rounded = [round(t) for t in temps_list if t < 24]
print(cool_rounded)     # [19, 22]

```

```python
# 2.1 rank with sorted
results = [("Aline", 85), ("Eric", 78), ("Diane", 91), ("Samuel", 92), ("Grace", 76)]
ranked = sorted(results, key=lambda r: r[1], reverse=True)
print(ranked[:3])   # [('Samuel', 92), ('Diane', 91), ('Aline', 85)]
```

Part 3 · NumPy Arrays 

```python
# 3.1 the speed test
import time
import numpy as np

n = 1_000_000

# Python loop on a plain list
python_list = list(range(n))
start = time.time()
result_loop = []
for x in python_list:
    result_loop.append(x * 2)
loop_time = time.time() - start

# NumPy: one vectorized operation
arr = np.arange(n)
start = time.time()
result_np = arr * 2
numpy_time = time.time() - start

print(f"Python loop: {loop_time:.4f} seconds")
print(f"NumPy:       {numpy_time:.6f} seconds")
print(f"NumPy was about {loop_time / numpy_time:.0f}x faster")

# COMMENT (the part that is graded):
# On my machine NumPy was roughly ___x faster than the Python loop
# because it runs the multiplication in compiled C code on a whole array
# at once, instead of one Python step per number.
```

Part 4 · Boolean Masking 

Exercise 4.1 — Hot days 


```python
# 4.1 hot days
hot_mask = temps >= 26
print("Hot days:", hot_mask.sum())                          # 18
print("Avg temp on hot days:", temps[hot_mask].mean())       # 27.11
```

```python

# 4.2 rain report
rainy = rain > 0
print("Rainy days:", rainy.sum())                            # 16
print("Total rainfall:", rain.sum())                         # 331.2 mm
print("Heaviest day:", rain.max())                           # 26.2 mm

```

```python
# 4.3 cross-array analysis
print("Avg temp rainy:", temps[rainy].mean())                # 24.14
print("Avg temp dry:  ", temps[~rainy].mean())               # 23.95
print("Hot AND rainy:", ((temps >= 26) & (rain > 0)).sum())  # 5
print("Humidity >= 70:", (hum >= 70).sum())                  # 16
```

Part 5 · Anomalies & Trends 
Exercise 5.1 — Anomaly detection

```python
# 5.1
anomaly = temps - temps.mean()
print(dates[anomaly > 3])
# ['2026-09-07' '2026-09-09' '2026-09-11' '2026-09-12' '2026-09-14'
#  '2026-09-16' '2026-09-17' '2026-09-19' '2026-09-22']   -> 9 days
print("Days > 3°C below mean:", (anomaly < -3).sum())       # 8

```

```python
# 5.2
first, last = temps[:30].mean(), temps[30:].mean()
print(f"First 30 days: {first:.2f} °C")   # 26.21
print(f"Last 30 days:  {last:.2f} °C")    # 21.78
# Kigali got COOLER: the average dropped by about 4.4 °C from September to October.
```

Part 6 · Reflection

1. What did NumPy hide, and why learn the loop first?
temps.mean() hides the loop, the running total, the counter and the division. NumPy runs those steps in fast compiled code. Learning the loop first means you understand what mean() actually does, so you can debug it, handle odd cases like missing or bad values, and tell when a loop is genuinely needed. Without that, NumPy is a black box.
