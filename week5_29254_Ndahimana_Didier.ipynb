```python
import numpy as np
import time
import pandas as pd


df = pd.read_excel("week5_kigali_weather.xlsx")

dates = np.array(df["date"].astype(str).tolist())
temps = np.array(df["temp_c"].astype(float).tolist())
rain  = np.array(df["rainfall_mm"].astype(float).tolist())
hum   = np.array(df["humidity_pct"].astype(float).tolist())

```

```python
#PART 1
#EXERCISE 1.1

temps_list = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1]

temps_f = [t * 9/5 + 32 for t in temps_list]
print(temps_f)
```

```python
#EXERCISE 1.2


hot = [t for t in temps_list if t >= 26]
print("Hot:", hot)

cool_rounded = [round(t) for t in temps_list if t < 24]
print("Cool rounded:", cool_rounded)
```

```python
# EXERCISE 1.3

raw = ["24.4", "25.9", "28.6", "19.3"]

raw_floats = [float(x) for x in raw]
average = sum(raw_floats) / len(raw_floats)
print("Average:", average)
```

```python
#PART 2

#EXERCISE 2.1

results = [("Aline", 85), ("Eric", 78), ("Diane", 91),
           ("Samuel", 92), ("Grace", 76)]

ranked = sorted(results, key=lambda s: s[1], reverse=True)
print("Top 3:", ranked[:3])
```

```python
#EXERCISE 2.2

rain_log = [(3, 0.0), (7, 12.5), (11, 3.2), (14, 26.2), (18, 8.9)]

wettest = max(rain_log, key=lambda pair: pair[1])
print("Wettest day:", wettest)
```

```python
#PART 3
#EXERCISE 3.1
numbers = list(range(1_000_000))
arr = np.array(numbers)


t0 = time.time()
doubled_py = [n * 2 for n in numbers]
py_time = time.time() - t0
print(f"Python: {py_time:.3f} s")


t0 = time.time()
doubled_np = arr * 2
np_time = time.time() - t0
print(f"NumPy: {np_time:.3f} s")


```

```python
#EXERCISE 3.2

print(f"Mean: {temps.mean():.2f} °C")
print(f"Max:  {temps.max():.2f} °C")
print(f"Min:  {temps.min():.2f} °C")
print(f"Std:  {temps.std():.2f} °C")
```

```python
# EXERCISE 3.3

position = np.argmax(temps)
print(f"Hottest day: {dates[position]}")
print(f"Temperature: {temps[position]:.2f} °C")
```

```python
#PART 4

#EXERCISE 4.1

hot_mask = temps >= 26
print("Hot days:", hot_mask.sum())
print(f"Avg temp on hot days: {temps[hot_mask].mean():.2f} °C")
```

```python
#EXERCISE  4.2

rainy_mask = rain > 0
print("Rainy days:", rainy_mask.sum())
print(f"Total rainfall: {rain.sum():.2f} mm")
print(f"Heaviest rainfall: {rain.max():.2f} mm")
```

```python
#EXERCISE 4.3

print(f"Avg temp on rainy days: {temps[rain > 0].mean():.2f} °C")
print(f"Avg temp on dry days:   {temps[rain == 0].mean():.2f} °C")

hot_and_rainy = (temps >= 26) & (rain > 0)
print("Hot & rainy days:", hot_and_rainy.sum())

humid_days = hum >= 70
print("Humid days (>= 70%):", humid_days.sum())
```

```python
# PART 5

#EXERCISE 5.1

anomaly = temps - temps.mean()

above_mask = anomaly > 3
print("Unusually hot days (dates):")
print(dates[above_mask])

below_mask = anomaly < -3
print("Unusually cool days count:", below_mask.sum())
```

```python
# EXERCISE 5.2
first_half  = temps[:30]
second_half = temps[30:]

print(f"First 30 days avg: {first_half.mean():.2f} °C")
print(f"Last 30 days avg:  {second_half.mean():.2f} °C")

```

```python
#BONUS 

pairs = list(zip(dates, rain))
top5 = sorted(pairs, key=lambda pair: pair[1], reverse=True)[:5]

print("Top 5 wettest days:")
for d, r in top5:
    print(f"  {d}  ->  {r:.2f} mm")
```
