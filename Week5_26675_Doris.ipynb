# Week 5 Lab

### Load data

```python
import numpy as np          # NumPy: fast math on arrays (we call it 'np' by convention)
import time                # used only for the speed test

with open("week5_kigali_weathers.csv") as f:      # open the file safely
    lines = f.readlines()[1:]                    # read all lines, skip the header row

rows  = [ln.strip().split(",") for ln in lines]  # each line -> list of 5 text pieces

dates = np.array([r[1] for r in rows])                # column 1: date (text)
temps = np.array([float(r[2]) for r in rows])         # column 2: temperature (convert text -> float)
rain  = np.array([float(r[3]) for r in rows])         # column 3: rainfall in mm
hum   = np.array([float(r[4]) for r in rows])         # column 4: humidity %

print("Days loaded:", temps.size)   # should be 60
```

# Part 1

```python
temps_list = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1]   
raw        = ["24.4", "25.9", "28.6", "19.3"]        
```

### 1.1 Fahrenheit 

```python
temps_f = [t * 9/5 + 32 for t in temps_list]   
print(temps_f)
```

### 1.2

```python
# (a) hot days: keep t only if t >= 26
hot = [t for t in temps_list if t >= 26]
print("hot:", hot)

# (b) cool days: keep t < 24, and round each one to the nearest whole number
cool_rounded = [round(t) for t in temps_list if t < 24]
print("cool_rounded:", cool_rounded)
```

###  1.3 

```python
raw_floats = [float(x) for x in raw]                 # "24.4" -> 24.4 for every item
print(raw_floats)
print("Average:", sum(raw_floats) / len(raw_floats))   # sum of all / how many
```

# Part 2 · lambda

### 2.1 

```python
results = [("Aline", 85), ("Eric", 78), ("Diane", 91), ("Samuel", 92), ("Grace", 76)]

# key tells sorted() WHAT to sort by: s[1] is the score (position 1 of each tuple)

ranked = sorted(results, key=lambda s: s[1], reverse=True)   
print("Top 3:", ranked[:3] )
```

###  2.2 

```python
rain_log = [(3, 0.0), (7, 12.5), (11, 3.2), (14, 26.2), (18, 8.9)]   # (day, rainfall_mm)

wettest = max(rain_log, key=lambda r: r[1])   # compare the pairs by rainfall (position 1)
print("Wettest day:", wettest)
```

# Part 3 

### 3.1 

```python
import numpy as np
import time

numbers = list(range(1_000_000))
arr = np.array(numbers)

t0 = time.time()
doubled = [n * 2 for n in numbers]          # pure Python: visits numbers one at a time
python_time = time.time() - t0
print(f"Python: {python_time:.4f} s")

t0 = time.time()
doubled = arr * 2                           # NumPy: vectorized, runs in fast compiled C
numpy_time = time.time() - t0
print(f"NumPy:  {numpy_time:.4f} s")

print(f"NumPy was roughly {python_time / numpy_time:.0f}x faster")

```

### 3.2 

```python
print(f"Mean: {temps.mean():.2f} °C")   # average
print(f"Max:  {temps.max():.1f} °C")     # hottest
print(f"Min:  {temps.min():.1f} °C")     # coldest
print(f"Std:  {temps.std():.2f} °C")     # spread: typical distance from the mean
```

### 3.3

```python
pos = np.argmax(temps)                 
print(f"Hottest day: {dates[pos]} at {temps[pos]} °C")   
```

# Part 4

### 4.1 

```python
hot_mask = temps >= 26                       

print("(a) Hot days:", hot_mask.sum())                            
print(f"(b) Average temp on hot days: {temps[hot_mask].mean():.2f} °C")   
```

### 4.2 

```python
rainy_mask = rain > 0                        # any rain at all

print("(a) Rainy days:", rainy_mask.sum())
print(f"(b) Total rainfall: {rain.sum():.1f} mm")
print(f"(c) Heaviest single day: {rain.max():.1f} mm  (on {dates[np.argmax(rain)]})")
```

### 4.3

```python
# (a) rainy vs dry average temperature
print(f"(a) Rainy days avg temp: {temps[rain > 0].mean():.2f} °C")
print(f"    Dry days avg temp:   {temps[rain == 0].mean():.2f} °C")

# (b) hot AND rainy: use & (not 'and'), with parentheses around each condition
hot_and_rainy = (temps >= 26) & (rain > 0)
print("(b) Hot AND rainy days:", hot_and_rainy.sum())

# (c) humid days
print("(c) Days with humidity >= 70:", (hum >= 70).sum())
```

# Part 5 ·

### 5.1 

```python
anomaly = temps - temps.mean()      # subtract the mean from EVERY element at once (vectorized)

# (a) dates more than 3 °C ABOVE the mean
print("(a) Unusually hot days:", dates[anomaly > 3])

# (b) count of days more than 3 °C BELOW the mean
print("(b) Unusually cold days:", (anomaly < -3).sum())
```

### 5.2

```python
first_half  = temps[:30]    # days 1-30
second_half = temps[30:]    # days 31-60

print(f"First 30 days average:  {first_half.mean():.2f} °C")
print(f"Last 30 days average:   {second_half.mean():.2f} °C")

# COMMENT: Kigali got COOLER — the second-half average is about 4 °C lower than the first half.
```

### Bonus

```python
pairs = list(zip(dates, rain))                                  # [(date, rain), ...]
top5  = sorted(pairs, key=lambda p: p[1], reverse=True)[:5]     # sort by rainfall, biggest first, keep 5

for d, mm in top5:                                              # loop only to print nicely
    print(f"{d}: {mm} mm")
```
