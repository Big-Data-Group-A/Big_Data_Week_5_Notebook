```python
import numpy as np
import time

arr = np.arange(1_000_000)

# Python loop
start = time.perf_counter()
loop_result = [x * 2 for x in arr]
loop_time = time.perf_counter() - start

# NumPy
start = time.perf_counter()
numpy_result = arr * 2
numpy_time = time.perf_counter() - start

speedup = loop_time / numpy_time

print(f"Python loop time: {loop_time:.6f} seconds")
print(f"NumPy time: {numpy_time:.6f} seconds")
print(f"NumPy was approximately {speedup:.2f} times faster.")
```

```python
import pandas as pd

df = pd.read_excel("/content/week5_kigali_weather.xlsx")

temps = df["temp_c"].to_numpy()
rain = df["rainfall_mm"].to_numpy()
hum = df["humidity_pct"].to_numpy()

rainy_mask = rain > 0
dry_mask = rain == 0

rainy_average = temps[rainy_mask].mean()
dry_average = temps[dry_mask].mean()

hot_and_rainy = ((temps >= 26) & (rain > 0)).sum()

humidity_70 = (hum >= 70).sum()

print(f"Average temperature on rainy days: {rainy_average:.2f} °C")
print(f"Average temperature on dry days: {dry_average:.2f} °C")
print(f"Hot AND rainy days: {hot_and_rainy}")
print(f"Days with humidity >= 70%: {humidity_70}")
```

```python
import pandas as pd

df = pd.read_excel("/content/week5_kigali_weather.xlsx")

dates = pd.to_datetime(
    df["date"],
    unit="D",
    origin="1899-12-30"
).dt.strftime("%Y-%m-%d").to_numpy()

temps = df["temp_c"].to_numpy()

anomaly = temps - temps.mean()

above_mask = anomaly > 3
below_mask = anomaly < -3

print("Dates more than 3 °C above the mean:")
print(dates[above_mask])

print(f"\nDays more than 3 °C below the mean: {below_mask.sum()}")
```

```python
import pandas as pd

df = pd.read_excel("/content/week5_kigali_weather.xlsx")

dates = pd.to_datetime(
    df["date"],
    unit="D",
    origin="1899-12-30"
).dt.strftime("%Y-%m-%d").to_numpy()

rain = df["rainfall_mm"].to_numpy()

wettest_5 = sorted(
    zip(dates, rain),
    key=lambda x: x[1],
    reverse=True
)[:5]

for date, rainfall in wettest_5:
    print(f"{date}: {rainfall:.1f} mm")
```
