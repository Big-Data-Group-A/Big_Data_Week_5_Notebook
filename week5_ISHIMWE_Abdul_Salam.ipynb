## Part 1 · List Comprehensions 

```python
temps = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1]
raw = ["24.4", "25.9", "28.6", "19.3"]
```

### Exercise 1.1 — Transform

```python
temps_f = [t * 9/5 + 32 for t in temps]
print(temps_f)
```

### Exercise 1.2 — Filter

```python
hot = [t for t in temps if t >= 26]
cool_rounded = [round(t) for t in temps if t < 24]

print(hot)
print(cool_rounded)
```

### Exercise 1.3 — Convert types

```python
floats = [float(x) for x in raw]
print(sum(floats) / len(floats))
```

## Part 2 · Lambda

### Exercise 2.1 — Rank with sorted

```python
results = [("Aline", 85), ("Eric", 78), ("Diane", 91), ("Samuel", 92), ("Grace", 76)]

ranked = sorted(results, key=lambda r: r[1], reverse=True)
print(ranked[:3])
```

### Exercise 2.2 — max with a key

```python
rain_log = [(3, 0.0), (7, 12.5), (11, 3.2), (14, 26.2), (18, 8.9)]

wettest = max(rain_log, key=lambda pair: pair[1])
print(wettest)
```

## Part 3 · NumPy Arrays

```python
import numpy as np

with open("week5_kigali_weather.csv") as f:
    lines = f.readlines()[1:]          # skip the header

rows = [line.strip().split(",") for line in lines]

dates = np.array([r[1] for r in rows])                 # text dates
temps = np.array([float(r[2]) for r in rows])          # temperatures
rain  = np.array([float(r[3]) for r in rows])          # rainfall
hum   = np.array([float(r[4]) for r in rows])          # humidity
```

### Exercise 3.1 — The speed test

```python
import time

big_list = list(range(1_000_000))
big_arr = np.array(big_list)

start = time.time()
doubled_list = [x * 2 for x in big_list]
list_time = time.time() - start

start = time.time()
doubled_arr = big_arr * 2
arr_time = time.time() - start

print(f"List: {list_time:.4f}s, NumPy: {arr_time:.4f}s")
print(f"NumPy was about {list_time / arr_time:.0f}x faster")
```

### Exercise 3.2 — Describe the temperatures

```python
print(f"Mean: {temps.mean():.2f} °C")
print(f"Max: {temps.max()} °C")
print(f"Min: {temps.min()} °C")
print(f"Std: {temps.std():.2f} °C")
```

### Exercise 3.3 — The hottest day

```python
position = np.argmax(temps)
print(f"Hottest day: {dates[position]} at {temps[position]} °C")
```

## Part 4 · Boolean Masking

### Exercise 4.1 — Hot days

```python
hot_mask = temps >= 26

print(f"Hot days: {hot_mask.sum()}")
print(f"Average temp on hot days: {temps[hot_mask].mean():.2f} °C")
```

### Exercise 4.2 — Rain report

```python
rainy_mask = rain > 0

print(f"Rainy days: {rainy_mask.sum()}")
print(f"Total rainfall: {rain.sum():.1f} mm")
print(f"Heaviest day: {rain.max()} mm")
```

### Exercise 4.3 — Cross-array analysis

```python
# (a) rainy vs dry average temperature
rainy_avg = temps[rain > 0].mean()
dry_avg = temps[rain == 0].mean()
print(f"Rainy days avg: {rainy_avg:.2f} °C, dry days avg: {dry_avg:.2f} °C")

# (b) hot AND rainy
hot_and_rainy = ((temps >= 26) & (rain > 0)).sum()
print(f"Hot and rainy days: {hot_and_rainy}")

# (c) humidity >= 70
print(f"Days with humidity >= 70: {(hum >= 70).sum()}")
```

## Part 5 · Anomalies & Trends

### Exercise 5.1: anomaly detection

```python
anomaly = temps - temps.mean()

# (a) dates more than 3 °C above the mean
print(dates[anomaly > 3])

# (b) count of days more than 3 °C below the mean
print((anomaly < -3).sum())
```

### Exercise 5.2 — Is the season changing

```python
first_half = temps[:30].mean()
second_half = temps[30:].mean()

print(f"First 30 days: {first_half:.2f} °C")
print(f"Last 30 days: {second_half:.2f} °C")
```

Kigali got  cooler in the second half because the temperature dropped to 21.78 °C

### Bonus — Top 5 wettest days

```python
pairs = zip(dates, rain)
wettest = sorted(pairs, key=lambda p: p[1], reverse=True)

for date, mm in wettest[:5]:
    print(f"{date}: {mm} mm")
```
