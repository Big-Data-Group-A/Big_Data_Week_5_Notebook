```python
df = pd.read_excel("week5_kigali_weather.xlsx")
temps = df["temp_c"].to_numpy()
rain = df["rainfall_mm"].to_numpy()
humidity = df["humidity_pct"].to_numpy()
```

```python
hot_mask = temps >= 26

print(f"(a) Hot days: {hot_mask.sum()}")
print(f"(b) Average temp on hot days: {temps[hot_mask].mean():.2f} °C")
```

```python
rainy_mask = rain > 0

print(f"(a) Rainy days: {rainy_mask.sum()}")
print(f"(b) Total rainfall: {rain.sum():.1f} mm")
print(f"(c) Heaviest single day: {rain.max():.1f} mm")
```

```python
# (a) rainy vs dry average temperature
print(f"(a) Rainy days avg temp: {temps[rain > 0].mean():.2f} °C")
print(f"    Dry days avg temp:   {temps[rain == 0].mean():.2f} °C")

# (b) hot AND rainy
print(f"(b) Hot and rainy days: {((temps >= 26) & (rain > 0)).sum()}")

# (c) humid days
print(f"(c) Days with humidity >= 70: {(humidity >= 70).sum()}")
```

```python
anomaly = temps - temps.mean()

# (a) dates more than 3 °C above the mean
print("(a) Days > 3 °C above mean:")
print(dates[anomaly > 3])

# (b) count of days more than 3 °C below the mean
print(f"(b) Days > 3 °C below mean: {(anomaly < -3).sum()}")
```

```python
first = temps[:30].mean()
last = temps[30:].mean()

print(f"First 30 days: {first:.2f} °C")
print(f"Last 30 days:  {last:.2f} °C")

# Kigali got COOLER: the average dropped by about 4.4 °C from September to October.
```

```python
wettest = sorted(zip(dates, rain), key=lambda p: p[1], reverse=True)[:5]

for d, mm in wettest:
    print(f"{d}: {mm} mm")
```
