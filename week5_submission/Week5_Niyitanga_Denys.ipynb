## Load the Kigali weather data

```python
import numpy as np
import time

with open("data/week5_kigali_weather.csv") as f:
    lines = f.readlines()[1:]

rows = [ln.strip().split(",") for ln in lines]
dates = np.array([r[1] for r in rows])
temps = np.array([float(r[2]) for r in rows])
rain = np.array([float(r[3]) for r in rows])
hum = np.array([float(r[4]) for r in rows])

print(f"Loaded {temps.size} days.")
print(f"Temperature mean: {temps.mean():.2f} °C")
```

## Part 1 · List Comprehensions

### Exercise 1.1 — Transform

```python
temps_part1 = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1]
temps_f = [t * 9/5 + 32 for t in temps_part1]
print(f"Fahrenheit: {temps_f}.:2f")
```

### Exercise 1.2 — Filter

```python
hot = [t for t in temps_part1 if t >= 26]
cool_rounded = [round(t) for t in temps_part1 if t < 24]
print(f"Hot: {hot}")
print(f"Cool rounded: {cool_rounded}")
```

### Exercise 1.3 — Convert types

```python
raw = ["24.4", "25.9", "28.6", "19.3"]
raw_floats = [float(t) for t in raw]
print(f"Floats: {raw_floats}")
print(f"Average: {sum(raw_floats) / len(raw_floats)}")
```

## Part 2 · lambda 

### Exercise 2.1 — Rank with `sorted`

```python
results = [("Aline", 85), ("Eric", 78), ("Diane", 91), ("Samuel", 92), ("Grace", 76)]
ranked = sorted(results, key=lambda x: x[1], reverse=True)
print(f"Top 3: {ranked[:3]}")
```

### Exercise 2.2 — max with a key

```python
rain_log = [(3, 0.0), (7, 12.5), (11, 3.2), (14, 26.2), (18, 8.9)]
wettest_day = max(rain_log, key=lambda x: x[1])
print(f"Wettest day: {wettest_day}")
```

## Part 3 · NumPy Arrays 

### Exercise 3.1 — The speed test
The exact speedup depends on the computer running the notebook. The Week 5 notes describe approximately 50× as a typical result and explain that 30×–200× can be normal.

```python
numbers = list(range(1_000_000))
arr = np.array(numbers)

t0 = time.time()
doubled_python = [n * 2 for n in numbers]
python_time = time.time() - t0

t0 = time.time()
doubled_numpy = arr * 2
numpy_time = time.time() - t0

speedup = python_time / numpy_time if numpy_time > 0 else float("inf")
print(f"Python: {python_time:.6f} s")
print(f"NumPy:  {numpy_time:.6f} s")
print(f"NumPy speedup: about {speedup:.1f}×")
```

### Exercise 3.2 — Describe the temperatures

```python
print(f"Mean: {temps.mean():.2f} °C")
print(f"Max: {temps.max():.2f} °C")
print(f"Min: {temps.min():.2f} °C")
print(f"Std: {temps.std():.2f} °C")
```

### Exercise 3.3 — The hottest day

```python
position = np.argmax(temps)
print(f"Hottest day: {dates[position]}")
print(f"Temperature: {temps[position]:.1f} °C")
```

## Part 4 · Boolean Masking 

### Exercise 4.1 — Hot days

```python
hot_mask = temps >= 26
print(f"Number of hot days: {hot_mask.sum()}")
print(f"Average hot-day temperature: {temps[hot_mask].mean():.2f} °C")
```

### Exercise 4.2 — Rain report

```python
rain_mask = rain > 0
print(f"Rainy days: {rain_mask.sum()}")
print(f"Total rainfall: {rain.sum():.1f} mm")
print(f"Heaviest single-day rainfall: {rain.max():.1f} mm")
```

### Exercise 4.3 — Cross-array analysis

```python
print(f"Average temperature on rainy days: {temps[rain > 0].mean():.2f} °C")
print(f"Average temperature on dry days: {temps[rain == 0].mean():.2f} °C")
hot_and_rainy = (temps >= 26) & (rain > 0)
print(f"Hot AND rainy days: {hot_and_rainy.sum()}")
print(f"Days with humidity >= 70%: {(hum >= 70).sum()}")
```

## Part 5 · Anomalies & Trends 

### Exercise 5.1 — Anomaly detection

```python
anomaly = temps - temps.mean()
above_mean_mask = anomaly > 3
below_mean_mask = anomaly < -3

print("Dates more than 3 °C above the mean:")
print(dates[above_mean_mask])
print(f"Days more than 3 °C below the mean: {below_mean_mask.sum()}")
```

### Exercise 5.2 — Is the season changing?

```python
first30_avg = temps[:30].mean()
last30_avg = temps[30:].mean()
print(f"First 30 days average: {first30_avg:.2f} °C")
print(f"Last 30 days average: {last30_avg:.2f} °C")
# Kigali got cooler because the last 30-day average is lower than the first 30-day average.
```

## Bonus — Top 5 wettest days

```python
wettest = sorted(zip(dates, rain), key=lambda x: x[1], reverse=True)
print("Top 5 wettest days:")
print(*wettest[:5], sep="\n")
```
