## Setup — load the data
*(If your lab notebook already has the loading cell, use that one instead.)*

```python
dates = np.arange('2025-01-01', '2025-03-01', dtype='datetime64[D]')
temps = np.linspace(19.0, 29.0, 60)
rain = np.round(np.maximum(np.sin(np.arange(60) * 0.9) * 18 + 8, 0.0), 1)
hum = np.round(np.linspace(55, 85, 60), 0)
print(f'Loaded {len(dates)} days of weather data')
```

## Part 1 · List Comprehensions

```python
import numpy as np
temps_list = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1]   
raw = ["24.4", "25.9", "28.6", "19.3"]               
```

### Exercise 1.1 — Transform

```python
fahrenheit = [t * 9/5 + 32 for t in temps_list]
print(fahrenheit)
```

### Exercise 1.2 — Filter

```python
hot = [t for t in temps_list if t >= 26]
cool_rounded = [round(t) for t in temps_list if t < 24]
print("hot:", hot)
print("cool_rounded:", cool_rounded)
```

### Exercise 1.3 — Convert types

```python
floats = [float(x) for x in raw]
print(floats)
print("Average:", sum(floats) / len(floats))
```

## Part 2 · lambda

### Exercise 2.1 — Rank with sorted

```python
results = [("Aline", 85), ("Eric", 78), ("Diane", 91), ("Samuel", 92), ("Grace", 76)]
ranked = sorted(results, key=lambda r: r[1], reverse=True)
print("Top 3:", ranked[:3])
```

### Exercise 2.2 — max with a key

```python
rain_log = [(3, 0.0), (7, 12.5), (11, 3.2), (14, 26.2), (18, 8.9)]
wettest = max(rain_log, key=lambda pair: pair[1])
print("Wettest day:", wettest)   # (day, rainfall_mm)
```

### Exercise 3.2 — Describe the temperatures

```python
print(f"Mean: {temps.mean():.2f} °C")
print(f"Max : {temps.max():.2f} °C")
print(f"Min : {temps.min():.2f} °C")
print(f"Std : {temps.std():.2f} °C")
# Checkpoint: mean should be 24.00 °C
```

### Exercise 3.3 — The hottest day

```python
pos = np.argmax(temps)
print(f"Hottest day: {dates[pos]} at {temps[pos]:.1f} °C")
```

## Part 4 · Boolean Masking

### Exercise 4.1 — Hot days

```python
hot_mask = temps >= 26
print("(a) Number of hot days:", hot_mask.sum())
print(f"(b) Average temp on hot days: {temps[hot_mask].mean():.2f} °C")
```

### Exercise 4.2 — Rain report

```python
rainy_mask = rain > 0
print("(a) Rainy days:", rainy_mask.sum())
print(f"(b) Total rainfall: {rain.sum():.1f} mm")
print(f"(c) Heaviest single-day rainfall: {rain.max():.1f} mm")
```

### Exercise 4.3 — Cross-array analysis

```python
rainy_mask = rain > 0
dry_mask = rain == 0
print(f"(a) Avg temp rainy days: {temps[rainy_mask].mean():.2f} °C")
print(f"    Avg temp dry days  : {temps[dry_mask].mean():.2f} °C")
print("(b) Hot AND rainy days:", ((temps >= 26) & (rain > 0)).sum())
print("(c) Days with humidity >= 70:", (hum >= 70).sum())
```

## Part 5 · Anomalies & Trends

### Exercise 5.1 — Anomaly detection

```python
anomaly = temps - temps.mean()
print("(a) Days > 3 °C above the mean:", dates[anomaly > 3])
print("(b) Days > 3 °C below the mean:", (anomaly < -3).sum())
```

### Exercise 5.2 — Is the season changing?

```python
first_half = temps[:30].mean()
second_half = temps[30:].mean()
print(f"First 30 days : {first_half:.2f} °C")
print(f"Last 30 days  : {second_half:.2f} °C")
print(f"Difference    : {second_half - first_half:+.2f} °C")
print("Kigali got", "warmer" if second_half > first_half else "cooler")
# COMMENT: Kigali got ___ (write warmer/cooler based on the output above),
# by about ___ °C between the first and second half of the period.
```

## Bonus (+3) — Top 5 wettest days

```python
pairs = sorted(zip(dates, rain), key=lambda p: p[1], reverse=True)
for d, r in pairs[:5]:      # a loop is allowed here, for printing the Bonus only
    print(f"{d}: {r:.1f} mm")
```

## Part 6 · Reflection
**1. What did NumPy hide from you, and why learn the loop version first?**
NumPy hides the iteration itself: the loop that visits every element, keeps a running total, and divides by the count. `temps.mean()` runs that work in fast compiled code, so I never see it. Learning the loop first mattered because it shows what the computation actually does. That helps me debug when results look wrong, understand why vectorized code is faster, and know what to do when no ready-made method exists.
