```python
import numpy as np
import time

n = 1_000_000

# Python loop
py_list = list(range(n))
start = time.perf_counter()
for i in range(n):
    py_list[i] = py_list[i] * 2
loop_time = time.perf_counter() - start

# NumPy vectorised
arr = np.arange(n)
start = time.perf_counter()
arr2 = arr * 2
numpy_time = time.perf_counter() - start

print(f"Python loop: {loop_time:.5f}s")
print(f"NumPy:       {numpy_time:.5f}s")
print(f"NumPy was about {loop_time / numpy_time:.1f}x faster")
```
