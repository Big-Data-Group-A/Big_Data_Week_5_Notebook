```python
#Exercise 3.3
pos = np.argmax(temps)
print(f"Hottest day: {dates[pos]} at {temps[pos]} C")
```

```python
#Exercise 4.1
hot = temps >= 26
print("Hot days:", hot.sum())
print("Average on hot days:", round(temps[hot].mean(), 2))
```

```python
#Exercise 4.2
rainy = rain > 0
print("Rainy days:", rainy.sum())
print("Total rainfall:", round(rain.sum(), 1), "mm")
print("Heaviest day:", rain.max(), "mm")
```

```python
#Exercise 4.3
# (a) mask from rain, data from temps
print("Avg temp, rainy days:", round(temps[rain > 0].mean(), 2))
print("Avg temp, dry days: ", round(temps[rain == 0].mean(), 2))
# (b) hot AND rainy (parentheses!)
print("Hot and rainy days:", ((temps >= 26) & (rain > 0)).sum())
# (c) humid days
print("Days with humidity >= 70:", (hum >= 70).sum())
```

```python
#Exercise 5.1
anomaly = temps - temps.mean()
print("Days more than 3 C above the mean:")
print(dates[anomaly > 3])
print("Days more than 3 C below the mean:", (anomaly < -3).sum())
```

```python
#Exercise 5.2
first = temps[:30].mean()
last = temps[30:].mean()
print(f"First 30 days: {first:.2f} C")
print(f"Last 30 days: {last:.2f} C")
# Kigali got COOLER: the average dropped from 26.21 C to 21.78 C (about 4.4 C lower).
```
