## Load the weather dataset

```python
import numpy as np
import pandas as pd

# Read the provided dataset
weather = pd.read_excel("week5_kigali_weather.xlsx")

# Convert Excel date serials into readable dates if needed
if pd.api.types.is_numeric_dtype(weather["date"]):
    weather["date"] = pd.to_datetime(weather["date"], unit="D", origin="1899-12-30")

dates = weather["date"].astype(str).to_numpy()
temps = weather["temp_c"].to_numpy(dtype=float)
rain = weather["rainfall_mm"].to_numpy(dtype=float)
hum = weather["humidity_pct"].to_numpy(dtype=float)

print(f"Records: {len(temps)}")
```

## Part 1 — List Comprehensions (6 marks)

```python
temps_part1 = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1]

# Exercise 1.1 — Convert Celsius to Fahrenheit
fahrenheit = [t * 9/5 + 32 for t in temps_part1]
print("Temperatures in Fahrenheit:", fahrenheit)

# Exercise 1.2(a) — Temperatures >= 26°C
hot = [t for t in temps_part1 if t >= 26]
print("Hot temperatures:", hot)

# Exercise 1.2(b) — Temperatures below 24°C, rounded
cool_rounded = [round(t) for t in temps_part1 if t < 24]
print("Cool temperatures (rounded):", cool_rounded)

# Exercise 1.3 — Convert strings to floats and calculate average
raw = ["24.4", "25.9", "28.6", "19.3"]
raw_floats = [float(x) for x in raw]
print("Floats:", raw_floats)
print("Average:", sum(raw_floats) / len(raw_floats))
```

## Part 2 — lambda (4 marks)

```python
# Exercise 2.1 — Sort results from highest to lowest and print top 3
results = [("Aline", 85), ("Eric", 78), ("Diane", 91), ("Samuel", 92), ("Grace", 76)]
ranked = sorted(results, key=lambda item: item[1], reverse=True)
print("Top 3:", ranked[:3])

# Exercise 2.2 — Find the wettest day
rain_log = [(3, 0.0), (7, 12.5), (11, 3.2), (14, 26.2), (18, 8.9)]
wettest = max(rain_log, key=lambda item: item[1])
print("Wettest day (day, rainfall_mm):", wettest)
```

## Part 4 — Boolean Masking (10 marks)

### Exercise 4.1 

```python
hot_mask = temps >= 26
print(f"Number of hot days (>= 26°C): {hot_mask.sum()}")
print(f"Average temperature on hot days: {temps[hot_mask].mean():.2f} °C")
```

### Exercise 4.2 

```python
rainy_mask = rain > 0
print(f"Number of rainy days: {rainy_mask.sum()}")
print(f"Total rainfall: {rain.sum():.1f} mm")
print(f"Heaviest single-day rainfall: {rain.max():.1f} mm")
```

### Exercise 4.3 

```python
dry_mask = rain == 0
print(f"Average temperature on rainy days: {temps[rainy_mask].mean():.2f} °C")
print(f"Average temperature on dry days: {temps[dry_mask].mean():.2f} °C")
print(f"Hot AND rainy days: ((temps >= 26) & (rain > 0)).sum() = {((temps >= 26) & (rain > 0)).sum()}")
print(f"Days with humidity >= 70%: {(hum >= 70).sum()}")
```

## Part 5 — Anomalies & Trends (7 marks)

### Exercise 5.1 

```python
anomaly = temps - temps.mean()
above_mask = anomaly > 3
below_mask = anomaly < -3
print("Dates more than 3°C above the mean:")
print(dates[above_mask])
print(f"Days more than 3°C below the mean: {below_mask.sum()}")
```

### Exercise 5.2 

```python
first30_avg = temps[:30].mean()
last30_avg = temps[30:].mean()
print(f"Average temperature, first 30 days: {first30_avg:.2f} °C")
print(f"Average temperature, last 30 days: {last30_avg:.2f} °C")
# Kigali was cooler in the last 30 days than in the first 30 days.
```

## Bonus 

```python
wettest_days = sorted(zip(dates, rain), key=lambda item: item[1], reverse=True)
print("Top 5 wettest days:")
for date, rainfall in wettest_days[:5]:
    print(f"{date}: {rainfall:.1f} mm")
```

## Part 6 — Reflection

**1. What did NumPy hide, and why learn loops first?**  
NumPy hides the repeated steps needed to process each value individually. An operation such as `temps.mean()` calculates the average of the whole array in one concise expression, while vectorized operations apply calculations to many values at once. Learning loops first is still important because loops help explain how repetition and accumulation work, making it easier to understand what the shorter NumPy operations are doing.

**2. One insight about Kigali's weather:**  
In this 60-day dataset, the average temperature fell from **26.21°C in the first 30 days** to **21.78°C in the last 30 days**. This indicates that the latter half of the recorded period was cooler. The dataset also records 16 rainy days and 331.2 mm of total rainfall.
