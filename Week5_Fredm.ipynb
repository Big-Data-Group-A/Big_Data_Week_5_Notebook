```python
import pandas as pd
import numpy as np
import time
```

```python
df = pd.read_csv("week5_kigali_weather.csv")
df.head()
```

```python
#PART 1
#1.1
temps = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1]
temp_f = [t * 9/5 + 32 for t in temps]
print(temp_f)
```

```python
#1.2
#a
hot = [t for t in temps if t >= 26]
print(hot)
```

```python
#b
cool_rounded = [round(t, 0) for t in temps if t < 24]
print(cool_rounded)
```

```python
#1.3
raw = ["24.4", "25.9", "28.6", "19.3"]
raw_f = [float(r) for r in raw]
print(sum(raw_f) / len(raw_f))
```

```python
#PART 2
#2.1
results = [("Aline", 85), ("Eric", 78), ("Diane", 91), ("Samuel", 92), ("Grace", 76)]
sorted_results = sorted(results, key=lambda r: r[1], reverse=True)
print(sorted_results[:3])
```

```python
#2.2
rain_log = [(3, 0.0), (7, 12.5), (11, 3.2), (14, 26.2), (18, 8.9)]
wettest_day = max(rain_log, key=lambda rain: rain[1])
print(wettest_day)
```

```python
#PART 3
#3.1
numbers = list(range(1000000))
arr = np.array(numbers)

#python loop
t0 = time.time()
d = [n * 2 for n in numbers]
print(f"Python: {time.time()-t0:.3f}s")

#Numpy
t0 = time.time()
d = arr * 2
print(f"Numpy: {time.time()-t0:.3f}s")

#Numpy was much faster, around 30x on my machine
```

```python
#3.2
temps = np.array(df.temp_c)
print(f"Mean : {round(temps.mean(), 1)}, Max : {temps.max()}, Min : {temps.min()}, Standard deviation : {temps.std()}")
```

```python
#3.3
hp = np.argmax(temps)
print(f"Date of the hottest day: {df.date[hp]}, temperature: {df.temp_c[hp]}")
```

```python
#PART 4
#4.1
#a
mask_t = temps >= 26
print(sum(mask_t))
```

```python
#b
print(temps[mask_t].mean())
```

```python
#4.2
rain = np.array(df.rainfall_mm)
mask_r = rain > 0
#a
print(f"{rain[mask_r].size} rainy days")
```

```python
#b
print(rain.sum())
```

```python
#c
print(rain.max())
```

```python
#4.3
#a
print(f"avg temp on rainy days: {round(temps[mask_r].mean(), 2)}")
print(f"avg temp on dry days: {round(temps[~mask_r].mean(), 2)}")
```

```python
#b
hot_and_rainy = (mask_t) & (mask_r)
print(hot_and_rainy.sum())
```

```python
#c
hum = np.array(df.humidity_pct)
mask_h = hum >= 70
print(f"{mask_h.sum()} days")
```

```python
#PART 5
#5.1
anomaly = temps - temps.mean()
#a
mask_a = anomaly > 3
print(df.date[mask_a])
```

```python
#b
mask_b = anomaly < -3
print(mask_b.sum())
```

```python
#5.2
print(temps[:30].mean())
print(temps[30:].mean())
#Kigali got cooler in October compared to September
```

```python
#BONUS
zipped = list(zip(df.date, rain))
sorted_zip = sorted(zipped, key=lambda z: z[1], reverse=True)
print(f"The 5 wettest days are: {sorted_zip[:5]}")
```
