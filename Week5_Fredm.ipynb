## Dataset Loading
Builds four NumPy arrays: `dates`, `temps`, `rain`, `hum`

```python
import numpy as np
import time

# Load using Week 3 file-reading + Week 5 comprehensions
with open("week5_kigali_weather.csv") as f:
    lines = f.readlines()[1:]          # skip header

rows  = [ln.strip().split(",") for ln in lines]

dates = [r[1] for r in rows]                       # list of date strings
temps = np.array([float(r[2]) for r in rows])      # °C
rain  = np.array([float(r[3]) for r in rows])      # mm
hum   = np.array([float(r[4]) for r in rows])      # %

print(f"Dataset loaded: {temps.size} days ready")
print(f"Date range    : {dates[0]}  →  {dates[-1]}")
```

## Part 1 · List Comprehensions  (6 marks)

```python
# Given data for Part 1
temps_list = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1]
raw        = ["24.4", "25.9", "28.6", "19.3"]
```

### Exercise 1.1 — Transform  (2 marks)
Convert every temperature to Fahrenheit (`t * 9/5 + 32`) with a **one-line** comprehension — no for block.

```python
# One-line comprehension: Celsius → Fahrenheit
temps_f = [t * 9/5 + 32 for t in temps_list]

print("Celsius   :", temps_list)
print("Fahrenheit:", [round(f, 2) for f in temps_f])
# Expected: [75.92, 78.62, 83.48, 66.74, 78.8, 71.78]
```

### Exercise 1.2 — Filter  (2 marks)
(a) `hot` = temperatures ≥ 26 · (b) `cool_rounded` = temperatures below 24, rounded with `round(t)`

```python
# (a) Hot temperatures — keep only >= 26
hot = [t for t in temps_list if t >= 26]
print("hot (>= 26°C)         :", hot)
# Expected: [28.6, 26.0]

# (b) Cool temperatures — keep only < 24, rounded
cool_rounded = [round(t) for t in temps_list if t < 24]
print("cool_rounded (< 24°C) :", cool_rounded)
# Expected: [19]
```

### Exercise 1.3 — Convert types  (2 marks)
CSV files give you strings. Convert `raw` to a list of floats with a comprehension, then print its average.

```python
# Convert list of strings → floats, then compute average
floats  = [float(x) for x in raw]
average = sum(floats) / len(floats)

print("String list :", raw)
print("Float list  :", floats)
print(f"Average     : {average:.2f}")
# Expected average: 24.55
```

## Part 2 · lambda  (4 marks)

### Exercise 2.1 — Rank with sorted  (2 marks)
Sort from highest to lowest score using `sorted(..., key=lambda ..., reverse=True)` and print the top 3.

```python
results = [("Aline", 85), ("Eric", 78), ("Diane", 91),
           ("Samuel", 92), ("Grace", 76)]

# Sort highest → lowest score with lambda
ranked = sorted(results, key=lambda s: s[1], reverse=True)

print("Top 3 students:")
for i, (name, score) in enumerate(ranked[:3], 1):
    print(f"  {i}. {name} — {score}")
# Expected: 1. Samuel-92  2. Diane-91  3. Aline-85
```

### Exercise 2.2 — max with a key  (2 marks)
Find the wettest day in **one line** with `max(..., key=lambda ...)`.

```python
rain_log = [(3, 0.0), (7, 12.5), (11, 3.2), (14, 26.2), (18, 8.9)]

# One line — find the pair with the highest rainfall
wettest = max(rain_log, key=lambda x: x[1])

print(f"Wettest day: Day {wettest[0]}  →  {wettest[1]} mm")
# Expected: Day 14 → 26.2 mm
```

## Part 3 · NumPy Arrays  (8 marks)

### Exercise 3.1 — The speed test  (2 marks)
Python loop vs `arr * 2` on one million numbers. Note roughly how many times faster NumPy was.

```python
numbers = list(range(1_000_000))
arr     = np.array(numbers)

# Pure Python
t0 = time.time()
doubled_py = [n * 2 for n in numbers]
py_time = time.time() - t0

# NumPy vectorized
t0 = time.time()
doubled_np = arr * 2
np_time = time.time() - t0

speedup = py_time / np_time if np_time > 0 else float('inf')

print(f"Python list comprehension : {py_time:.4f} s")
print(f"NumPy vectorized          : {np_time:.4f} s")
print(f"NumPy was ~{speedup:.0f}x faster on this machine")
# NumPy runs the loop in compiled C — typically ~50x faster
```

### Exercise 3.2 — Describe the temperatures  (3 marks)
Using array methods only: print mean, max, min, and std of `temps`.  
**Checkpoint: the mean is 24.00 °C.**

## Part 4 · Boolean Masking  (10 marks)

### Exercise 4.1 — Hot days  (3 marks)
A hot day is ≥ 26 °C. Using masks: (a) how many hot days? (b) the average temperature on those days?

```python
# Boolean mask for hot days
hot_mask = temps >= 26

print(f"(a) Number of hot days (>= 26 °C) : {hot_mask.sum()}")
# Expected: 20

print(f"(b) Average temperature on hot days: {temps[hot_mask].mean():.2f} °C")
# Expected: 27.15 °C
```

### Exercise 4.2 — Rain report  (3 marks)
Using masks on `rain`: (a) rainy days, (b) total rainfall, (c) heaviest single-day rainfall.

```python
# Boolean mask for rainy days
rainy_mask = rain > 0

print(f"(a) Number of rainy days         : {rainy_mask.sum()}")
# Expected: 20

print(f"(b) Total rainfall over 60 days  : {rain.sum():.1f} mm")
# Expected: 314.6 mm

print(f"(c) Heaviest single-day rainfall : {rain.max():.1f} mm")
# Expected: 26.2 mm  (2026-09-07)
```

### Exercise 4.3 — Cross-array analysis  (4 marks)
(a) Avg temp rainy vs dry days · (b) hot AND rainy days · (c) days with humidity ≥ 70

```python
# (a) Average temperature — rainy days vs dry days
#     mask from rain array, data from temps array
avg_temp_rainy = temps[rain > 0].mean()
avg_temp_dry   = temps[rain == 0].mean()

print(f"(a) Avg temp on rainy days : {avg_temp_rainy:.2f} °C")  # → 25.02
print(f"    Avg temp on dry days   : {avg_temp_dry:.2f} °C")    # → 23.36
print(f"    → Dry days are {'warmer' if avg_temp_dry > avg_temp_rainy else 'cooler'} than rainy days")

# (b) Hot AND rainy — use & with parentheses around each condition!
hot_and_rainy_mask = (temps >= 26) & (rain > 0)
print(f"\n(b) Days that were hot (>= 26 °C) AND rainy : {hot_and_rainy_mask.sum()}")
# Expected: 9

# (c) Days with humidity >= 70 %
print(f"(c) Days with humidity >= 70 %           : {(hum >= 70).sum()}")
# Expected: 20
```

## Part 5 · Anomalies & Trends  (7 marks)

### Exercise 5.1 — Anomaly detection  (4 marks)
Compute `anomaly = temps - temps.mean()` (vectorized).  
(a) Print the DATES of all days more than 3 °C above the mean.  
(b) Count the days more than 3 °C below the mean.

```python
# Vectorized anomaly — one line, no loop
anomaly = temps - temps.mean()

print(f"Mean temperature : {temps.mean():.2f} °C")

# (a) Dates of days more than 3 °C ABOVE the mean
above_mask  = anomaly > 3
above_dates = [dates[i] for i in np.where(above_mask)[0]]

print(f"\n(a) Days more than 3 °C above the mean ({above_mask.sum()} days):")
for i in np.where(above_mask)[0]:
    print(f"    {dates[i]}  →  {temps[i]:.1f} °C  (anomaly: +{anomaly[i]:.2f} °C)")
# Expected dates: 2026-09-07, 09-12, 09-14, 09-16, 09-17, 09-19, 09-22

# (b) Count days more than 3 °C BELOW the mean
below_count = (anomaly < -3).sum()
print(f"\n(b) Days more than 3 °C below the mean : {below_count}")
# Expected: 5
```

### Exercise 5.2 — Is the season changing?  (3 marks)
Compare `temps[:30]` (September) with `temps[30:]` (October). State in a comment whether Kigali got warmer or cooler.

```python
# NumPy slicing — no loops needed
first_half_avg  = temps[:30].mean()   # September
second_half_avg = temps[30:].mean()   # October

print(f"First 30 days  (Sep 01 – Sep 30) average : {first_half_avg:.2f} °C")
# Expected: 26.21 °C

print(f"Last  30 days  (Oct 01 – Oct 30) average : {second_half_avg:.2f} °C")
# Expected: 21.78 °C

print(f"Difference : {second_half_avg - first_half_avg:.2f} °C")
# Expected: -4.43 °C

# Comment: Kigali got COOLER moving from September into October.
# The temperature dropped by ~4.4 °C — a clear seasonal cooling trend.
direction = "COOLER" if second_half_avg < first_half_avg else "WARMER"
print(f"→ Kigali got {direction} in the second half of the period.")
```

## Bonus (+3) — Top 5 Wettest Days
Pair dates with rainfall using `zip(dates, rain)`, rank with `sorted(..., key=lambda ..., reverse=True)`, print the 5 wettest days.

```python
# zip dates with rain → sort by rainfall descending → slice top 5
date_rain_pairs = list(zip(dates, rain))

top5_wettest = sorted(date_rain_pairs, key=lambda x: x[1], reverse=True)[:5]

print("Top 5 Wettest Days:")
for rank, (date, rainfall) in enumerate(top5_wettest, 1):
    print(f"  {rank}. {date}  →  {rainfall:.1f} mm")
# Expected:
# 1. 2026-09-07  →  26.2 mm
# 2. 2026-09-14  →  25.1 mm
# 3. 2026-09-21  →  24.0 mm
# 4. 2026-09-05  →  22.9 mm
# 5. 2026-09-28  →  22.9 mm
```

### Question 1
> *In Week 3 you needed ~15 lines to average a column; today it was `temps.mean()`. What did NumPy hide from you — and why was it still important to learn the loop version first?*

**Answer:**

NumPy hid the iteration, accumulation, and type-management logic we wrote manually in Week 3. Behind `temps.mean()`, NumPy runs a highly optimised C loop that sums every element and divides by the count — work that required us 4–6 lines of Python code. It also hides memory layout: a NumPy array stores all values as contiguous `float64` bytes in memory, which is exactly why the CPU can process them so much faster than a Python list of separate objects.

It was still important to learn the loop version first for three reasons:
1. **Conceptual clarity** — knowing *what* `temps.mean()` does under the hood (iterate, sum, divide) makes us better at debugging unexpected results and understanding edge cases.
2. **Error detection** — the manual loop forced us to handle empty lists, non-numeric values, and division by zero explicitly. NumPy handles these quietly; the loop made the problems visible and teachable.
3. **Portability** — not every environment ships with NumPy. Understanding the algorithm means we can implement it in any language or framework.
