## Setup — load the data into four NumPy arrays

```python
import numpy as np

FILE = "week5_kigali_weather.csv"

# columns: 0 day · 1 date · 2 temp_c · 3 rainfall_mm · 4 humidity_pct
dates = np.loadtxt(FILE, delimiter=",", skiprows=1, usecols=1, dtype=str)
temps, rain, hum = np.genfromtxt(FILE, delimiter=",", skip_header=1,
                                 usecols=(2, 3, 4), unpack=True)

print("Loaded", len(dates), "days:", dates[0], "to", dates[-1])
print("temps[:5] =", temps[:5])
```

## Part 3 · NumPy Arrays (8 marks)

### Exercise 3.1 — The speed test (2 marks)

```python
import time

N = 1_000_000
py_list = list(range(N))
np_arr = np.arange(N)

t0 = time.perf_counter()
doubled_list = []                            # pure Python loop
for x in py_list:
    doubled_list.append(x * 2)
py_time = time.perf_counter() - t0

t0 = time.perf_counter()
doubled_arr = np_arr * 2                     # vectorized NumPy
np_time = time.perf_counter() - t0

print(f"Python loop: {py_time:.4f} s")
print(f"NumPy      : {np_time:.4f} s")
print(f"NumPy was about {py_time / np_time:.0f}x faster")

# NumPy was roughly 10x faster in my run (update this to the number YOUR machine prints; it varies;
# see the number printed above). It runs the loop in optimized C instead of Python.
```

## Part 4 · Boolean Masking (10 marks)

### Exercise 4.1 — Hot days (3 marks)

```python
hot_mask = temps >= 26

print(f"(a) Number of hot days: {hot_mask.sum()}")
print(f"(b) Average temperature on hot days: {temps[hot_mask].mean():.2f} °C")
```

### Exercise 4.2 — Rain report (3 marks)

```python
rainy_mask = rain > 0

print(f"(a) Rainy days: {rainy_mask.sum()}")
print(f"(b) Total rainfall over 60 days: {rain.sum():.1f} mm")
print(f"(c) Heaviest single-day rainfall: {rain.max():.1f} mm")
```

### Exercise 4.3 — Cross-array analysis (4 marks)

```python
# (a) mask from rain, data from temps
print(f"(a) Avg temp on rainy days: {temps[rainy_mask].mean():.2f} °C")
print(f"    Avg temp on dry days  : {temps[~rainy_mask].mean():.2f} °C")

# (b) hot AND rainy  (parentheses around each condition)
print(f"(b) Hot AND rainy days: {((temps >= 26) & (rain > 0)).sum()}")

# (c) humidity >= 70
print(f"(c) Days with humidity >= 70%: {(hum >= 70).sum()}")
```

## Part 5 · Anomalies & Trends (7 marks)

### Exercise 5.1 — Anomaly detection (4 marks)

```python
anomaly = temps - temps.mean()

warm_mask = anomaly > 3
print("(a) Days more than 3 °C above the mean:")
print(dates[warm_mask])

print(f"(b) Days more than 3 °C below the mean: {(anomaly < -3).sum()}")
```

### Exercise 5.2 — Is the season changing? (3 marks)

```python
first30 = temps[:30].mean()
last30 = temps[30:].mean()
print(f"First 30 days average: {first30:.2f} °C")
print(f"Last 30 days average : {last30:.2f} °C")
print(f"Change: {last30 - first30:+.2f} °C")

# Conclusion: Kigali got COOLER. The last 30 days average about 4.4 °C lower
# than the first 30, so the season cooled as September turned into October.
```

## Bonus (+3) — Top 5 wettest days

```python
pairs = sorted(zip(dates, rain), key=lambda p: p[1], reverse=True)
for d, mm in pairs[:5]:          # the only loop allowed: printing the bonus
    print(f"{d}: {mm} mm")
```

## Part 6 · Reflection

**1. What did NumPy hide from you — and why was it still important to learn the loop version first?**

`temps.mean()` hides the whole mechanism: creating a running total, walking through every element one at a time, counting them, and dividing at the end. It also hides that the work is done by fast compiled C code rather than by Python's slower loop, which is why the speed test showed such a large gap. Learning the loop version first matters because I now understand *what* `mean()` is actually computing, so I can trust it, debug it when something looks wrong (for example a missing value or wrong data type), and write my own logic when no built-in method exists. Without the loop version, `mean()` would just be a magic word; with it, I know what the magic does.

**2. One insight about Kigali's weather**

The season clearly cooled and got more humid: the first 30 days averaged about 26.2 °C, while the last 30 averaged about 21.8 °C, and average humidity rose from about 58% to 66%. Rain fell on only 16 of 60 days, yet those days delivered 331 mm in total, and the five wettest days were spread across the period (7, 14, 21 and 28 Sept, plus 5 Sept) rather than bunched together. Rainy and dry days had almost the same average temperature (about 24.1 °C vs 23.9 °C), so rain by itself did not explain the cooling. The drop in temperature looks like a seasonal shift rather than an effect of individual rainy days, and the wettest days fell in the warmer first half (183 mm in the first 30 days vs 148 mm in the last 30).
