```python
#Exercise 3.1 — The speed test
import time
import numpy as np

arr = np.arange(1_000_000)

# Python loop
start = time.time()
result_loop = [x * 2 for x in arr]
loop_time = time.time() - start

# NumPy
start = time.time()
result_numpy = arr * 2
numpy_time = time.time() - start

print(f"Python loop: {loop_time:.6f} seconds")
print(f"NumPy: {numpy_time:.6f} seconds")
print(f"NumPy was approximately {loop_time / numpy_time:.1f} times faster.")
```

# Part 4 · Boolean Masking

```python
#Exercise 4.1
import pandas as pd
import numpy as np

data = pd.read_csv("week5_kigali_weather.csv")
temps = data["temp_c"].values

# Exercise 4.1 — Hot days
hot_mask = temps >= 26
num_hot_days = np.sum(hot_mask)
avg_hot_temp = np.mean(temps[hot_mask])
print(f"Number of hot days: {num_hot_days}")
print(f"Average temperature on hot days: {avg_hot_temp:.2f} °C")
```

```python
#Exercise 4.2
import pandas as pd
import numpy as np

data = pd.read_csv("week5_kigali_weather.csv")
rain = data["rainfall_mm"].values

# Exercise 4.2 — Rain report
rainy_mask = rain > 0
num_rainy_days = np.sum(rainy_mask)
total_rainfall = np.sum(rain)
max_rainfall = np.max(rain)
print(f"Number of rainy days: {num_rainy_days}")
print(f"Total rainfall over 60 days: {total_rainfall:.1f} mm")
print(f"Heaviest single-day rainfall: {max_rainfall:.1f} mm")
```

```python
#Exercise 4.3
import pandas as pd
import numpy as np

data = pd.read_csv("week5_kigali_weather.csv")
temps = data["temp_c"].values
rain = data["rainfall_mm"].values
humidity = data["humidity_pct"].values
# Cross-array analysis
rainy_mask = rain > 0
hot_mask = temps >= 26
humid_mask = humidity >= 70
avg_temp_rainy = np.mean(temps[rainy_mask])
avg_temp_dry = np.mean(temps[~rainy_mask])
hot_and_rainy = np.sum(hot_mask & rainy_mask)
num_humid_days = np.sum(humid_mask)
print(f"Average temp on rainy days: {avg_temp_rainy:.2f} °C")
print(f"Average temp on dry days: {avg_temp_dry:.2f} °C")
print(f"Number of hot AND rainy days: {hot_and_rainy}")
print(f"Number of days with humidity ≥ 70: {num_humid_days}")
```

# Part 5 · Anomalies & Trends

```python
#Exercise 5.1
import pandas as pd
import numpy as np

data = pd.read_csv("week5_kigali_weather.csv")
dates = data["date"].values
temps = data["temp_c"].values
#Anomaly detection
mean_temp = np.mean(temps)
anomaly = temps - mean_temp
# (a) Days > 3 °C above mean
above_mask = anomaly > 3
print("Dates more than 3 °C above mean:", dates[above_mask])
# (b) Days > 3 °C below mean
below_mask = anomaly < -3
count_below = np.sum(below_mask)
print("Number of days more than 3 °C below mean:", count_below)
```

```python
#Exercise 5.2
import pandas as pd
import numpy as np

data = pd.read_csv("week5_kigali_weather.csv")
temps = data["temp_c"].values

#— Season change
first30_avg = np.mean(temps[:30])
last30_avg = np.mean(temps[30:])
print(f"Average temperature (first 30 days): {first30_avg:.2f} °C")
print(f"Average temperature (last 30 days): {last30_avg:.2f} °C")

if last30_avg > first30_avg:
    print("Kigali got warmer.")
else:
    print("Kigali got cooler.")
```
