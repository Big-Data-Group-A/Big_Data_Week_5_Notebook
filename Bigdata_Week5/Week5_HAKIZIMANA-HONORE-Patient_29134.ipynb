```python
import numpy as np
import pandas as pd

# Access the Excel file
file_path = "week5_kigali_weather-LAB3.xlsx"
df = pd.read_excel(file_path, sheet_name="week5_kigali_weather")

# Extract columns as NumPy arrays
temps = df["temp_c"].to_numpy(dtype=float)
rain = df["rainfall_mm"].to_numpy(dtype=float)
humidity = df["humidity_pct"].to_numpy(dtype=float)

# Convert Excel dates into readable dates
date_column = df["date"]

if pd.api.types.is_numeric_dtype(date_column):
    dates = pd.to_datetime(
        date_column, unit="D", origin="1899-12-30"
    ).dt.strftime("%Y-%m-%d").to_numpy()
else:
    dates = pd.to_datetime(date_column).dt.strftime(
        "%Y-%m-%d"
    ).to_numpy()

print("Weather data loaded successfully.")
print(f"Number of records: {len(temps)}")
```

```python
#Exercises 4.1
# Create a Boolean mask for hot days
hot_mask = temps >= 26

# (a) Count hot days
print(f"Number of hot days: {hot_mask.sum()}")

# (b) Average temperature on hot days
print(
    f"Average temperature on hot days: "
    f"{temps[hot_mask].mean():.2f} °C"
)
```

```python
#Exercises 4.2
# Create a Boolean mask for rainy days
rain_mask = rain > 0

# (a) Number of rainy days
print(f"Number of rainy days: {rain_mask.sum()}")

# (b) Total rainfall over 60 days
print(f"Total rainfall: {rain.sum():.1f} mm")

# (c) Heaviest single-day rainfall
print(f"Heaviest rainfall: {rain.max():.1f} mm")
```

```python
#Exercises 4.3
# (a) Average temperature on rainy versus dry days
rain_mask = rain > 0
dry_mask = rain == 0

print(
    f"Average temperature on rainy days: "
    f"{temps[rain_mask].mean():.2f} °C"
)

print(
    f"Average temperature on dry days: "
    f"{temps[dry_mask].mean():.2f} °C"
)

# (b) Count days that were hot AND rainy
hot_and_rainy = (temps >= 26) & (rain > 0)

print(f"Number of hot and rainy days: {hot_and_rainy.sum()}")

# (c) Count days with humidity >= 70%
humid_mask = humidity >= 70

print(f"Days with humidity >= 70%: {humid_mask.sum()}")
```

```python
#Exercise 5.1
# Calculate temperature anomalies using vectorized subtraction
anomaly = temps - temps.mean()

# (a) Dates more than 3 °C above the mean
above_mask = anomaly > 3

print("Dates more than 3 °C above the mean:")
print(dates[above_mask])

# (b) Count days more than 3 °C below the mean
below_mask = anomaly < -3

print(
    f"Days more than 3 °C below the mean: "
    f"{below_mask.sum()}"
)
```

```python
#Exercises 5.2
# Calculate the average temperature for each period
first_avg = temps[:30].mean()
last_avg = temps[30:].mean()

print(f"Average temperature during first 30 days: {first_avg:.2f} °C")
print(f"Average temperature during last 30 days: {last_avg:.2f} °C")
print(f"Temperature change: {last_avg - first_avg:.2f} °C")

# Kigali got cooler because the average temperature
# in the last 30 days was lower than in the first 30 days.
```

```python
# Pair dates with rainfall and sort from wettest to driest
wettest_days = sorted(
    zip(dates, rain),
    key=lambda item: item[1],
    reverse=True
)

# Print the top 5 wettest days
print("Top 5 wettest days:")

for date, rainfall in wettest_days[:5]:
    print(f"{date}: {rainfall:.1f} mm")
```
