```python
import numpy as np

data = np.genfromtxt('week5_kigali_weather.csv', delimiter=',', skip_header=1, dtype=None, encoding='utf-8')

dates = np.array([row[1] for row in data])
temps = np.array([row[2] for row in data], dtype=float)
rain  = np.array([row[3] for row in data], dtype=float)
hum   = np.array([row[4] for row in data], dtype=float)

print('Dataset loaded successfully!')
print(f'Number of days: {len(dates)}')
print(f'Date range: {dates[0]} → {dates[-1]}')
print(f'First 5 temps: {temps[:5]}')
```

```python
temps_p1 = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1]
raw      = ["24.4", "25.9", "28.6", "19.3"]
```

```python
fahrenheit = [t * 9/5 + 32 for t in temps_p1]

print("Temperatures in Celsius:   ", temps_p1)
print("Temperatures in Fahrenheit:", [round(f, 2) for f in fahrenheit])
```

```python
hot = [t for t in temps_p1 if t >= 26]
cool_rounded = [round(t) for t in temps_p1 if t < 24]

print("(a) Hot temperatures (>= 26°C):       ", hot)
print("(b) Cool temperatures (< 24°C, rounded):", cool_rounded)
```

```python
raw_floats = [float(t) for t in raw]
average = sum(raw_floats) / len(raw_floats)

print("Raw strings: ", raw)
print("As floats:   ", raw_floats)
print(f"Average:      {average:.2f}")
```

```python
results = [("Aline", 85), ("Eric", 78), ("Diane", 91), ("Samuel", 92), ("Grace", 76)]

ranked = sorted(results, key=lambda x: x[1], reverse=True)

print("Top 3 students:")
print(ranked[:3])
```

```python
rain_log = [(3, 0.0), (7, 12.5), (11, 3.2), (14, 26.2), (18, 8.9)]

wettest_day = max(rain_log, key=lambda x: x[1])

print(f"Wettest day: Day {wettest_day[0]} with {wettest_day[1]} mm of rainfall")
```

```python
import time

N = 1_000_000
big_list = list(range(N))
big_arr  = np.arange(N, dtype=float)

start = time.time()
result_loop = [x * 2 for x in big_list]
loop_time = time.time() - start

start = time.time()
result_numpy = big_arr * 2
numpy_time = time.time() - start

speedup = loop_time / numpy_time

print(f"Python list comprehension: {loop_time:.4f} s")
print(f"NumPy vectorized:          {numpy_time:.4f} s")
print(f"NumPy was approximately {speedup:.1f}x faster")
```

```python
print(f"Mean temperature : {temps.mean():.2f} °C")
print(f"Max  temperature : {temps.max():.2f} °C")
print(f"Min  temperature : {temps.min():.2f} °C")
print(f"Std  deviation   : {temps.std():.2f} °C")
```

```python
position = np.argmax(temps)

print(f"Hottest day  : {dates[position]}")
print(f"Temperature  : {temps[position]:.1f} °C")
```

```python
hot_mask = temps >= 26
num_hot  = hot_mask.sum()
avg_hot  = temps[hot_mask].mean()

print(f"(a) Number of hot days (>= 26°C): {num_hot}")
print(f"(b) Average temperature on hot days: {avg_hot:.2f} °C")
```

```python
rain_mask  = rain > 0
num_rainy  = rain_mask.sum()
total_rain = rain.sum()
max_rain   = rain.max()

print(f"(a) Number of rainy days        : {num_rainy}")
print(f"(b) Total rainfall (60 days)    : {total_rain:.1f} mm")
print(f"(c) Heaviest single-day rainfall: {max_rain:.1f} mm")
```

```python
avg_temp_rainy = temps[rain > 0].mean()
avg_temp_dry   = temps[rain == 0].mean()

print(f"(a) Avg temp on rainy days : {avg_temp_rainy:.2f} °C")
print(f"    Avg temp on dry days   : {avg_temp_dry:.2f} °C")

hot_and_rainy = ((temps >= 26) & (rain > 0)).sum()
print(f"\n(b) Days that were hot AND rainy: {hot_and_rainy}")

high_humidity_days = (hum >= 70).sum()
print(f"\n(c) Days with humidity >= 70%   : {high_humidity_days}")
```

```python
anomaly = temps - temps.mean()

above_mask  = anomaly > 3
dates_above = dates[above_mask]

print("(a) Dates with temp > 3°C above the mean:")
print(dates_above)

below_count = (anomaly < -3).sum()
print(f"\n(b) Number of days more than 3°C below the mean: {below_count}")
```

```python
avg_first_30 = temps[:30].mean()
avg_last_30  = temps[30:].mean()

print(f"Average temperature — first 30 days (Sep): {avg_first_30:.2f} °C")
print(f"Average temperature — last  30 days (Oct): {avg_last_30:.2f} °C")
```

```python
ranked_rain = sorted(zip(dates, rain), key=lambda x: x[1], reverse=True)

print("Top 5 wettest days:")
for date, mm in ranked_rain[:5]:
    print(f"  {date}  →  {mm:.1f} mm")
```
