```python
import pandas as pd
import numpy as np

df = pd.read_excel("week5_kigali_weather.xlsx")

dates = pd.to_datetime(df["date"], unit="D", origin="1899-12-30").to_numpy()
temps = df["temp_c"].to_numpy()
rain = df["rainfall_mm"].to_numpy()
hum = df["humidity_pct"].to_numpy()
```

```python
# Exercise 3.2

print(f"Mean: {temps.mean():.2f} °C")
print(f"Max: {temps.max():.2f} °C")
print(f"Min: {temps.min():.2f} °C")
print(f"Std: {temps.std():.2f} °C")
```

```python
# Exercise 3.3

position = np.argmax(temps)

print("Hottest day:", dates[position])
print("Temperature:", temps[position], "°C")
```

```python
# Exercise 4.1

hot_mask = temps >= 26

print("Hot days:", hot_mask.sum())
print("Average hot temperature:", temps[hot_mask].mean())
```

```python
# Exercise 4.2

rain_mask = rain > 0

print("Rainy days:", rain_mask.sum())
print("Total rainfall:", rain.sum(), "mm")
print("Heaviest rainfall:", rain.max(), "mm")
```

```python
# Exercise 4.3

rainy_mask = rain > 0
dry_mask = rain == 0

print("Average rainy temperature:", temps[rainy_mask].mean())
print("Average dry temperature:", temps[dry_mask].mean())

hot_rainy = (temps >= 26) & (rain > 0)
print("Hot and rainy days:", hot_rainy.sum())

humid_days = hum >= 70
print("Days with humidity >= 70%:", humid_days.sum())
```

```python
# Exercise 5.1

anomaly = temps - temps.mean()

above_mask = anomaly > 3
below_mask = anomaly < -3

print("Dates more than 3°C above mean:")
print(dates[above_mask])

print("Days more than 3°C below mean:", below_mask.sum())
```

```python
# Exercise 5.2

first_30_avg = temps[:30].mean()
last_30_avg = temps[30:].mean()

print(f"First 30 days average: {first_30_avg:.2f} °C")
print(f"Last 30 days average: {last_30_avg:.2f} °C")

# Kigali got warmer if the last 30 days have a higher average temperature.
```

# Kigali got cooler.

```python
# Top 5 wettest days

wettest = sorted(zip(dates, rain), key=lambda x: x[1], reverse=True)

for day, rainfall in wettest[:5]:
    print(day, rainfall, "mm")
```
